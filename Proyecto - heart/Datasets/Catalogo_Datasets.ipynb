# 📦 Catálogo de Datasets — Proyectos del Módulo DLCD

**Uso:** Asignación individual de datasets para los proyectos de UD1, UD2 y UD3

---

## Criterios de selección

Los 15 datasets de este catálogo cumplen los siguientes criterios:

- **Origen:** Kaggle o UCI Machine Learning Repository. Todos son de acceso libre.
- **Usabilidad:** puntuación 10/10 en Kaggle (o equivalente en UCI).
- **Tamaño:** fichero CSV inferior a 100 MB.
- **Tipos de variables:** mezcla de variables numéricas y categóricas en todos los casos.
- **Calidad:** presencia confirmada de valores ausentes y/o valores extremos (outliers).
- **Dominio:** campos variados para evitar que los estudiantes compartan resultados.

> 📌 **Nota sobre datasets grandes:** los datasets con más de 30.000 filas incluyen una indicación de muestra recomendada (~15.000 filas estratificadas). El alumno puede trabajar con el dataset completo si su equipo lo permite, pero la muestra garantiza tiempos de ejecución razonables.

## 1 · Stroke Prediction Dataset

🔗 **Enlace:** https://www.kaggle.com/datasets/fedesoriano/stroke-prediction-dataset  
🏥 **Dominio:** Neurología / Medicina preventiva  
⭐ **Usabilidad Kaggle:** 10.0

### Descripción

Conjunto de datos clínicos con 11 características de pacientes, usado para predecir la probabilidad de sufrir un ictus. Incluye datos demográficos (edad, género, estado civil), indicadores de salud (hipertensión, enfermedades cardíacas, glucosa media, IMC) y hábitos de vida (tabaquismo, tipo de trabajo, residencia). Es un dataset con notable desbalanceo de clases (solo el 4,9% de casos positivos), lo que añade un reto adicional al análisis.

### Ficha técnica

| Característica | Valor |
|---------------|-------|
| **Filas** | 5.110 |
| **Columnas** | 12 (11 features + id) |
| **Variables numéricas** | age, avg_glucose_level, bmi |
| **Variables categóricas** | gender, hypertension, heart_disease, ever_married, work_type, Residence_type, smoking_status |
| **Variable objetivo** | stroke (0 = no, 1 = sí) — clasificación binaria |
| **Valores ausentes** | bmi: ~201 valores nulos (~3,9%) |
| **Outliers** | avg_glucose_level (cola derecha pronunciada), bmi |
| **Tamaño CSV** | ~312 KB |
| **Muestra recomendada** | Dataset completo (tamaño manejable) |

### Aspectos de interés para el proyecto

- La columna `bmi` tiene valores ausentes reales (MAR probable: tienden a faltar en jóvenes).
- `avg_glucose_level` tiene una distribución bimodal con sesgo positivo fuerte: candidata a transformación.
- `smoking_status` incluye la categoría 'Unknown' que puede tratarse como missing.
- El fuerte desbalanceo de clases (4.9% positivos) es un hallazgo clave a documentar.

## 2 · Cardiovascular Disease Dataset

🔗 **Enlace:** https://www.kaggle.com/datasets/sulianova/cardiovascular-disease-dataset  
❤️ **Dominio:** Cardiología / Epidemiología  
⭐ **Usabilidad Kaggle:** 10.0

### Descripción

Dataset de 70.000 pacientes con datos de examen médico para predecir la presencia de enfermedad cardiovascular. Incluye medidas objetivas (tensión arterial sistólica y diastólica, altura, peso, colesterol, glucosa), datos subjetivos (tabaquismo, consumo de alcohol, actividad física) y características demográficas.

### Ficha técnica

| Característica | Valor |
|---------------|-------|
| **Filas** | 70.000 |
| **Columnas** | 12 |
| **Variables numéricas** | age (en días), height, weight, ap_hi, ap_lo |
| **Variables categóricas** | gender, cholesterol (1-3), gluc (1-3), smoke, alco, active |
| **Variable objetivo** | cardio (0 = no, 1 = sí) — clasificación binaria |
| **Valores ausentes** | Sin NaN explícitos; los valores incorrectos actúan como ausentes |
| **Outliers / errores** | height ≤ 0, ap_hi/ap_lo negativos o extremos, weight < 10 |
| **Tamaño CSV** | ~3,0 MB |
| **Muestra recomendada** | 15.000 filas (estratificada por cardio) |

### Aspectos de interés para el proyecto

- Los valores ausentes están codificados como **valores biológicamente imposibles** (altura = 0, presión arterial negativa), no como NaN. Este patrón es muy común en datos médicos reales y requiere detección por reglas de dominio.
- La conversión de `age` (en días) a años es una transformación necesaria.
- Excelente dataset para calcular el IMC como variable derivada: `weight / (height/100)²`.
