### 2.1 Estadísticos descriptivos

Calcula los estadísticos descriptivos básicos para las variables numéricas y la distribución de frecuencias para las categóricas.

````python
# ── Variables numéricas ───────────────────────────────────────────────────
# TU CÓDIGO AQUÍ
#quiero a continuación imprimir mas cosas sin que desaparezca df.describe().T, por lo que lo guardo en una variable
numeric_summary = df.describe().T

# Conteo de valores en variables categóricas
print(numeric_summary)
for var in categorical_vars:
    print(f'Conteo de valores en {var}:')
    print(df[var].value_counts())
    print()

````

````python
#Diagramas para las variables categóricas, 
#relacionadas con la variable objetivo 'stroke', stroke 0 y 1 en la misma barra de cada variable categórica
for var in categorical_vars:
    if var != 'stroke':
        pd.crosstab(df[var], df['stroke'], normalize='index').plot(kind='bar', stacked=True)
        plt.title(f'Relación entre {var} y stroke')
        plt.ylabel('Proporción')
        plt.xlabel(var)
        plt.show()
````

### 2.2 Distribución de las variables numéricas

Visualiza la distribución de cada variable numérica. Identifica si son simétricas, sesgadas o bimodales.

````python
# TU CÓDIGO AQUÍ: histograma + boxplot para cada variable numérica
# Puedes usar la función diagnostico_visual() de los cuadernos de práctica
# o construir tus propios gráficos.

# Ejemplo de estructura:
cols_num = numeric_vars.copy()  # Crea una copia de la lista
cols_num.remove('id')  # Excluye la columna 'id' si no es relevante
fig, axes = plt.subplots(len(cols_num), 2, figsize=(12, 4*len(cols_num)))
for i, col in enumerate(cols_num):
    #set the title for each row
    axes[i,0].set_title(f'Histograma de {col}')
    axes[i,0].axvline(df[col].mean(), color='red', linestyle='dashed', linewidth=1, label='Media')
    axes[i,0].axvline(df[col].median(), color='green', linestyle='dashed', linewidth=1, label='Mediana')
    axes[i,0].legend
    axes[i,1].set_title(f'Boxplot de {col}')
    axes[i,0].hist(df[col].dropna(), bins=30, color='skyblue', edgecolor='black')
    axes[i,1].boxplot(df[col].dropna())
plt.tight_layout()
plt.show()
````

### 3.2 Outliers

Detecta valores atípicos en las variables numéricas usando el método IQR. Clasifícalos como errores o como valores extremos legítimos.

````python
# TU CÓDIGO AQUÍ: detección de outliers por IQR para cada variable numérica
# Puedes usar la función detectar_outliers_iqr() de los cuadernos de práctica

# Ejemplo:
for col in cols_num:
     Q1 = df[col].quantile(0.25)
     Q3 = df[col].quantile(0.75)
     IQR = Q3 - Q1
     n_out = ((df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)).sum()
     print(f'{col}: {n_out} outliers ({n_out/len(df)*100:.1f}%)')
````

````python
# TU CÓDIGO AQUÍ: visualiza los outliers de las 2-3 variables más afectadas
fig, axes = plt.subplots(1, 1, figsize=(14, 4))
sns.boxplot(data=df, y='avg_glucose_level',ax=axes)
plt.tight_layout()
plt.show()
````

````python
# Gráfico de glucosa vs edad, coloreando cada punto según la variable categórica 'stroke'
# Pintar en último lugar los puntos con stroke=1 para que se vean mejor

df_sorted = df.sort_values('stroke')
sc1 = plt.scatter(df_sorted['age'], df_sorted['avg_glucose_level'], c=df_sorted['stroke'], cmap='coolwarm', alpha=0.7)
plt.xlabel('Edad')
plt.ylabel('Nivel de Glucosa Promedio')
plt.title('Relación entre Edad y Nivel de Glucosa por Stroke')
plt.colorbar(sc1, label='Stroke')
plt.show()

# Gráfico de bmi vs edad, coloreando cada punto según la variable categórica 'stroke' (excluyendo NaN en bmi)

df_sorted = df.sort_values('stroke')
mask = df_sorted['bmi'].notna()
sc2 = plt.scatter(df_sorted.loc[mask, 'age'], df_sorted.loc[mask, 'bmi'], c=df_sorted.loc[mask, 'stroke'], cmap='coolwarm', alpha=0.7)
plt.xlabel('Edad')
plt.ylabel('BMI')
plt.title('Relación entre Edad y BMI por Stroke')
plt.colorbar(sc2, label='Stroke')
plt.show()
````
